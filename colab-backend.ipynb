```python
# Set up ngrok authentication (replace with your token)
# Get your token from: https://dashboard.ngrok.com/get-started/your-authtoken
import getpass
ngrok_token = getpass.getpass('Enter your ngrok auth token: ')
!ngrok config add-authtoken {ngrok_token}
```

```python
# Create environment file
env_content = """"
NODE_ENV=production
PORT=3000
MINECRAFT_SERVER_PATH=./minecraft-server
ADMIN_USERNAME=admin
ADMIN_PASSWORD=colab_server_2025
""""

with open('.env', 'w') as f:
    f.write(env_content.strip())

print("Environment file created!")
print("Default credentials:")
print("Username: admin")
print("Password: colab_server_2025")
```

```python
# Start ngrok tunnel in background
import subprocess
import time

# Start ngrok
ngrok_process = subprocess.Popen(['ngrok', 'http', '3000'], 
                               stdout=subprocess.PIPE, 
                               stderr=subprocess.PIPE)

# Wait a moment for ngrok to start
time.sleep(3)

# Get ngrok URL
try:
    result = subprocess.run(['curl', '-s', 'http://localhost:4040/api/tunnels'], 
                          capture_output=True, text=True)
    import json
    tunnels = json.loads(result.stdout)
    public_url = tunnels['tunnels'][0]['public_url']
    print(f"🚀 Your backend is now accessible at: {public_url}")
    print(f"📝 Update your frontend to use this URL")
except:
    print("❌ Could not get ngrok URL. Check ngrok status manually.")
    print("Run: !curl http://localhost:4040/api/tunnels")
```

```python
# Start the Minecraft control panel backend
print("🎮 Starting Minecraft Server Control Panel Backend...")
print("📊 The server will be available at the ngrok URL above")
print("⚠️  Keep this Colab tab open to keep the server running")
print("💡 You can access the frontend at: https://yehiah.github.io/minecraft-control-panel")

# Start the server
!npm start
```

- **Storage**: Files are saved to Google Drive for persistence
